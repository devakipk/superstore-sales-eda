Step 3 :- Data Visualization

```python
# plotting multiple graphs using subplot

# Setting figure size
plt.figure(figsize=(18, 12))

# Chart 1: Sales by Category
plt.subplot(2, 3, 1)

cat_sales = df.groupby("Category")["Sales"].sum()

plt.bar(cat_sales.index, cat_sales.values, color="blue")
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")


# Chart 2: Sales by Sub-Category
plt.subplot(2, 3, 2)

subcat_sales = df.groupby("Sub-Category")["Sales"].sum()

plt.bar(subcat_sales.index, subcat_sales.values, color="red")
plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Total Sales")

# Chart 3: Sales by Segment
plt.subplot(2, 3, 3)

segment_sales = df.groupby("Segment")["Sales"].sum()

plt.bar(segment_sales.index, segment_sales.values, color="green")
plt.title("Sales by Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")


# Chart 4: Sales by Region
plt.subplot(2, 3, 4)

region_sales = df.groupby("Region")["Sales"].sum()

plt.bar(region_sales.index, region_sales.values, color="blue")
plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Total Sales")


# Chart 5: Top 5 Cities by Sales
plt.subplot(2, 3, 5)

city_sales = df.groupby("City")["Sales"].sum()
top_cities = city_sales.sort_values(ascending=False).head(5)

plt.bar(top_cities.index,top_cities.values,color="red")

plt.title("Top 5 Cities by Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")


# Chart 6: Top 10 States by Sales
plt.subplot(2, 3, 6)

state_sales = df.groupby("State")["Sales"].sum()
top_states = state_sales.sort_values(ascending=False).head(10)

plt.bar(top_states.index, top_states.values, color="green")
plt.title("Top 10 States by Sales")
plt.xlabel("State")
plt.ylabel("Total Sales")

# Adjusting spacing
plt.tight_layout()

# Displaying charts
plt.show()
```

Step 4 :- outlier detection

```python

# finding outlier using IQR
columns = ["Sales", "Profit", "Discount"]

for col in columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\nColumn: {col}")
    print("Lower Limit:", lower_limit)
    print("Upper Limit:", upper_limit)
    print("Total Outliers:", len(outliers))
```

```python
# plotting box to detect outlier

plt.figure(figsize=(12, 5))

for i, col in enumerate(
    ["Sales", "Profit", "Discount"], 1
):
    plt.subplot(1, 3, i)
    sns.boxplot(y=df[col])
    plt.title(f"Outliers in {col}")

plt.tight_layout()
plt.show()
```

```python

columns = ["Sales", "Profit", "Discount"]

for col in columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_limit) |
        (df[col] > upper_limit)
    ]

    print(f"\n{col} Outliers:")
    print(outliers[col].describe())

    print("\nFirst 10 Outlier Values:")
    print(outliers[col].head(10).to_list())
```

Step 5 :- Business Insights

A. Which Category Generates the Highest Sales and Lowest sales?

```python
category_sales = (
    df.groupby("Category")["Sales"]
    .sum()
    .sort_values(ascending=False)
)

print(category_sales)

print(
    "\nHighest Sales Category:",
    category_sales.idxmax()
)

print(
    "\nLowest Sales Category:",
    category_sales.idxmin()
)
```

B. Which Region Generates the Highest Profit and Lower Profit?

```python

region_profit = (
    df.groupby("Region")["Profit"]
    .sum()
    .sort_values(ascending=False)
)

print(region_profit)

print(
    "\nHighest Profit Region:",
    region_profit.idxmax()
)

print(
    "\nLowest Profit Region:",
    region_profit.idxmin()
)
```

C. Which Sub-Category Has the Highest and Lowest Profit?

```python

subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values()
)

print(subcategory_profit)

print(
    "\nHighest Profit Sub-Category:",
    subcategory_profit.idxmax()
)
print(
    "\nLowest Profit Sub-Category:",
    subcategory_profit.idxmin()
)
```

D. What Is the Overall Profit Margin?

```python
total_sales = df["Sales"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Profit Margin:", round(profit_margin, 2), "%")
```
